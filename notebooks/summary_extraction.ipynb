```python
import pandas as pd
```

```python
df = pd.read_csv("../data/intelligence_items_gold.csv")
```

```python
published_dates = pd.to_datetime(df["published_at"], errors="coerce", utc=True)
week_start = published_dates.dt.normalize() - pd.to_timedelta(
    (published_dates.dt.weekday + 2) % 7, unit="D"
)

df["week_id"] = week_start.dt.strftime("%Y-%m-%d")
```

```python
last_10_weeks_df = df.loc[
    week_start >= week_start.max() - pd.Timedelta(weeks=9),
    ["summary", "week_id"],
].copy()

last_10_weeks_df
```

```python
last_10_weeks_df.to_csv("last_10_weeks.csv", index=False)
```
